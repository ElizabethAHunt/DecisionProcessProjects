```python
import pandas as pd
import numpy as np
```

```python
#load data into dataframe
DF_0 = pd.read_csv("GeneratedBusinessData.csv")
```

```python
#check for duplicates
print(DF_0[DF_0.duplicated()])
```

```python
#drop duplicates if they exist
DF_0.drop_duplicates(inplace=True)
```

```python
#GROUP BY state and CALCULATE mean, median, min & max for each numeric value for each state -- store as new dataframe

cols = [
    "Total Long-term Debt",
    "Total Equity",
    "Debt to Equity",
    "Total Liabilities",
    "Total Revenue",
    "Profit Margin"
]

DF_0[cols] = DF_0[cols].apply(pd.to_numeric)

DF_1 = DF_0.groupby("Business State")[cols].agg(
    ["mean", "median", "min", "max"]
)
```

```python
DF_1.head()
```

```python
#IF debt to equity ratio is negative, remove and save as “DF_2”
DF_2 = DF_0[DF_0[('Debt to Equity')] < 0]
DF_2.head()
```

```python
#ADD new column “Debt-to-income” WHERE IF total_revenue is 0, return NaN, ELSE divide Long-Term Debt by Revenue
DF_3 = pd.DataFrame({
    'Debt to Income Ratio': np.where(
        DF_0['Total Revenue'] == 0,
        np.nan,
        DF_0['Total Long-term Debt']/DF_0['Total Revenue']
    )
})
DF_3.head()
```

```python
#concat debt to income ratio to original dataframe
DF_Final = pd.concat([DF_0.reset_index(drop=True),DF_3.reset_index(drop=True)],axis=1)
DF_Final.head()
```

```python
#export to csv
DF_Final.to_csv(r'Output.csv')
DF_1.to_csv(r'Stats.csv')
```
